### Agora vamos ver a bibloteca matplotlib

```python
import matplotlib.pyplot as plt
from pylab import plot, show, legend, title, xlabel, ylabel, axis
import numpy as np
```

```python
plt.plot(['A', 'B', 'C', 'D'], [10, 20, 15, 25])
plt.ylabel('Valores')
plt.xlabel('Categorias')
plt.title('Gráfico de Linhas')
plt.show()
plt.close()
```

```python
mes = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vlr_1 = np.random.randint(2000, 5000, size=12)
vlr_2 = np.random.randint(4000, 6000, size=12)

plt.plot(mes, vlr_1, mes, vlr_2)
plt.ylabel('valor comissao')
plt.xlabel('Mes')
plt.title('Comparativo de comissão ano 2026')
plt.legend(['carlos', 'jose'])
plt.axis()
plt.show()
plt.close()
```

```python
linguagens = ['Python', 'C#', 'Java', 'R']
vlrs = [40, 50, 80, 20]
explode = [0,0,0.1,0]
cores = ['green', 'red', 'orange','yellow']


plt.pie(vlrs, explode=explode, labels=linguagens, colors=cores, autopct='%1.1f%%', startangle=140, shadow=True)
plt.axis('equal')
plt.show()
```
